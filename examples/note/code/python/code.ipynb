```python
from pathlib import Path

from matplotlib import pyplot as plt
from numpy import linspace

output_dir = Path("../../output/results/python")
output_dir.mkdir(parents=True, exist_ok=True)
x = linspace(-1, 1, 200)
fig, ax = plt.subplots()
ax.plot(x, x**2)
ax.set_xlabel("x")
ax.set_ylabel("x²")
fig.savefig(output_dir / "quadratic.png", bbox_inches="tight")
plt.close(fig)
```
